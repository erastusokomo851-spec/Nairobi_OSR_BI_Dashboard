## Objective

This notebook develops Key Performance Indicators (KPIs) that summarize the financial performance of Nairobi City County's Own Source Revenue collection. These KPIs will later be integrated into an interactive Business Intelligence dashboard.

```python
# ==========================================================
# IMPORT LIBRARIES
# ==========================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from matplotlib.ticker import FuncFormatter

pd.set_option("display.max_columns", None)
pd.options.display.float_format = "{:,.2f}".format

sns.set_style("whitegrid")
```

```python
# ==========================================================
# LOAD DATA
# ==========================================================

file_path = "../data/processed/OSR2025JUNE30_Clean.xlsx"

df = pd.read_excel(file_path)

df.head()
```

```python
# ==========================================================
# CREATE CONTRIBUTION PERCENTAGE
# ==========================================================

df["Contribution (%)"] = (
    df["Total"] / df["Total"].sum()
) * 100

df.head()
```

## KPI 1 — Total Own Source Revenue

```python
total_revenue = df["Total"].sum()

print(f"Total Revenue: KES {total_revenue:,.2f}")
```

## KPI 2 — Number of Revenue Streams

```python
total_streams = df["Revenue_Stream"].nunique()

print(f"Revenue Streams: {total_streams}")
```

## KPI 3 — Average Revenue per Stream

```python
average_revenue = df["Total"].mean()

print(f"Average Revenue: KES {average_revenue:,.2f}")
```

## KPI 4 — Highest Revenue Stream

```python
highest = df.loc[df["Total"].idxmax()]

highest

print(f"Highest Revenue Stream : {highest['Revenue_Stream']}")

print(f"Revenue : KES {highest['Total']:,.2f}")
```

## KPI 5 — Lowest Revenue Stream

```python
lowest = df.loc[df["Total"].idxmin()]

print(f"Lowest Revenue Stream : {lowest['Revenue_Stream']}")

print(f"Revenue : KES {lowest['Total']:,.2f}")
```

## KPI 6 — Revenue Concentration

```python
top5 = (
    df
    .sort_values("Total", ascending=False)
    .head(5)
)

top5_percentage = (
    top5["Total"].sum()
    /
    total_revenue
) * 100

print(f"Top Five Streams Contribute {top5_percentage:.2f}% of Total Revenue")
```

## KPI 7 — Quarter Performance

```python
quarter_totals = df[
    ["Q1","Q2","Q3","Q4"]
].sum()

quarter_totals
```

```python
best_quarter = quarter_totals.idxmax()

best_value = quarter_totals.max()

print(best_quarter)

print(best_value)
```

## KPI 8 — Quarter Contribution

```python
quarter_summary = pd.DataFrame({
    "Revenue": quarter_totals
})

quarter_summary["Contribution (%)"] = (
    quarter_summary["Revenue"]
    /
    quarter_summary["Revenue"].sum()
) * 100

quarter_summary
```

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Revenue Streams",
        "Average Revenue",
        "Highest Revenue Stream",
        "Lowest Revenue Stream",
        "Top 5 Contribution (%)"
    ],
    "Value": [
        f"KES {total_revenue:,.2f}",
        total_streams,
        f"KES {average_revenue:,.2f}",
        highest["Revenue_Stream"],
        lowest["Revenue_Stream"],
        f"{top5_percentage:.2f}%"
    ]
})

kpi_summary
```

## KPI Function 

```python
# ==========================================================
# KPI FUNCTIONS
# ==========================================================

def total_revenue(df):
    return df["Total"].sum()


def average_revenue(df):
    return df["Total"].mean()


def revenue_streams(df):
    return df["Revenue_Stream"].nunique()


def highest_stream(df):
    return df.loc[df["Total"].idxmax()]


def lowest_stream(df):
    return df.loc[df["Total"].idxmin()]

total_revenue(df)
```

## Executive KPIs

```python
# ==========================================================
# EXECUTIVE KPIs
# ==========================================================

total_osr = total_revenue(df)

avg_osr = average_revenue(df)

streams = revenue_streams(df)

highest = highest_stream(df)

lowest = lowest_stream(df)
```

## Create executive scorecard 

```python
scorecard = pd.DataFrame({

"KPI":[

"Total Revenue",

"Revenue Streams",

"Average Revenue",

"Highest Revenue Stream",

"Highest Revenue",

"Lowest Revenue Stream",

"Lowest Revenue"

],

"Value":[

total_osr,

streams,

avg_osr,

highest["Revenue_Stream"],

highest["Total"],

lowest["Revenue_Stream"],

lowest["Total"]

]

})

scorecard
```

## Revenue Ranking

```python
df["Rank"] = (

df["Total"]

.rank(

ascending=False,

method="dense"

)

.astype(int)

)

df.sort_values("Rank").head(10)
```

## Perfomance Classification

```python
conditions = [

df["Contribution (%)"] >= 15,

df["Contribution (%)"] >= 5,

df["Contribution (%)"] < 5

]

choices = [

"High",

"Medium",

"Low"

]

df["Performance"] = np.select(

conditions,

choices,

default="Low"

)

df.head()
```

## Revenue dependancy 

```python
top5 = (

df

.sort_values("Total",ascending=False)

.head(5)

)

dependency = (

top5["Total"].sum()

/

total_osr

)*100

print(dependency)
```

## Quarter perfomance index 

```python
quarter_totals = df[

["Q1","Q2","Q3","Q4"]

].sum()

quarter_index = pd.DataFrame({

"Quarter":quarter_totals.index,

"Revenue":quarter_totals.values

})

quarter_index["Rank"] = (

quarter_index["Revenue"]

.rank(

ascending=False,

method="dense"

)

.astype(int)

)

quarter_index
```

## Executive Summary 

```python
print("="*60)

print("EXECUTIVE REVENUE SUMMARY")

print("="*60)

print(f"Total Revenue : KES {total_osr:,.2f}")

print(f"Revenue Streams : {streams}")

print(f"Average Revenue : KES {avg_osr:,.2f}")

print(f"Highest Revenue Stream : {highest['Revenue_Stream']}")

print(f"Lowest Revenue Stream : {lowest['Revenue_Stream']}")

print(f"Top Five Revenue Dependency : {dependency:.2f}%")

print(f"Best Performing Quarter : {quarter_index.loc[quarter_index['Rank']==1,'Quarter'].iloc[0]}")

print("="*60)
```
