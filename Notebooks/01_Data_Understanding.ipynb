# 5. Initial Data Cleaning

The dataset contains summary rows (subtotal and grand total) where the Revenue Stream field is blank (`NaN`). These rows are intended for reporting purposes and do not represent individual revenue streams. They are removed to prevent double counting during analysis and dashboard development.

```python
# ======================================================
# CHECK MISSING REVENUE STREAMS
# ======================================================

missing_revenue_streams = df["Revenue Stream"].isna().sum()

print(f"Rows with missing Revenue Stream: {missing_revenue_streams}")
```

```python
# ======================================================
# REMOVE SUMMARY ROWS
# ======================================================

df = df.dropna(subset=["Revenue Stream"])

print("Updated Dataset Shape:", df.shape)
```

```python
df["Revenue Stream"].isna().sum()
```

```python
# ======================================================
# RESET INDEX
# ======================================================

df.reset_index(drop=True, inplace=True)

df.head(16)
df.tail()
```

```python
# Create a cleaned copy of the raw dataset
df_clean = df.copy()

# Remove summary rows
df_clean = df_clean.dropna(subset=["Revenue Stream"])

# Reset the index
df_clean.reset_index(drop=True, inplace=True)
```

# 6. Data Quality Assessment

This section evaluates the quality of the cleaned dataset by checking for duplicate records, missing values, incorrect data types, and consistency of quarterly revenue figures. The objective is to ensure that the dataset is suitable for analysis and dashboard development.

```python
# ======================================================
# DUPLICATE RECORDS
# ======================================================

duplicates = df_clean.duplicated().sum()

print("="*50)
print("DUPLICATE RECORDS")
print("="*50)
print(f"Duplicate Records Found: {duplicates}")
```

```python
# ======================================================
# MISSING VALUES
# ======================================================

missing = df_clean.isnull().sum()

print("="*50)
print("MISSING VALUES")
print("="*50)

print(missing)
```

```python
# ======================================================
# VALIDATE TOTAL COLUMN
# ======================================================

df_clean["Calculated_Total"] = (
    df_clean["Q1"] +
    df_clean["Q2"] +
    df_clean["Q3"] +
    df_clean["Q4"]
)

comparison = (
    df_clean["Calculated_Total"] ==
    df_clean["Total"]
)

print("All totals correct:", comparison.all())
```
